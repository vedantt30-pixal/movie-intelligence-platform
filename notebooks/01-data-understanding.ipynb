## Baseline Recommendation Model

The baseline model uses TF-IDF vectorization on a combined movie
feature representation consisting of:

- Genres
- Keywords
- Top cast
- Director
- Movie overview

Cosine similarity is used to calculate movie-to-movie similarity.

Dataset:
- 4,803 movies
- 10,000 TF-IDF features

```python
def genre_consistency(title, n=10):
    if title not in indices:
        return None

    # Get the first matching movie index
    idx = indices[title]

    if isinstance(idx, (pd.Series, list, np.ndarray)):
        idx = idx[0]

    recommendations = get_recommendations(title, n)

    original_genres = set(df.loc[idx, "genres_list"])

    scores = []

    for movie_title in recommendations["title"]:
        movie_idx = indices[movie_title]

        if isinstance(movie_idx, (pd.Series, list, np.ndarray)):
            movie_idx = movie_idx[0]

        recommended_genres = set(
            df.loc[movie_idx, "genres_list"]
        )

        if len(original_genres) == 0:
            continue

        overlap = (
            len(original_genres & recommended_genres)
            / len(original_genres)
        )

        scores.append(overlap)

    return sum(scores) / len(scores) if scores else 0
```

```python
import ast
import pandas as pd

def get_genres(movie_index):
    genres = df.iloc[movie_index]["genres"]

    if pd.isna(genres):
        return set()

    try:
        genre_list = ast.literal_eval(genres)
        return {g["name"] for g in genre_list}
    except:
        return set()


def genre_consistency(movie_title, recommendations):
    matches = df[df["title"].str.lower() == movie_title.lower()]

    if matches.empty:
        return 0

    movie_index = matches.index[0]
    source_genres = get_genres(movie_index)

    if not source_genres:
        return 0

    scores = []

    for title in recommendations["title"]:
        rec_matches = df[df["title"].str.lower() == title.lower()]

        if rec_matches.empty:
            continue

        rec_index = rec_matches.index[0]
        rec_genres = get_genres(rec_index)

        if not rec_genres:
            continue

        intersection = source_genres.intersection(rec_genres)
        union = source_genres.union(rec_genres)

        if union:
            scores.append(len(intersection) / len(union))

    return sum(scores) / len(scores) if scores else 0
```

```python
test_movies = ["Avatar", "The Dark Knight", "Interstellar"]

comparison_results = []

for movie in test_movies:

    tfidf_recs = get_recommendations(movie, 10)
    semantic_recs = get_semantic_recommendations(movie, 10)

    tfidf_score = genre_consistency(movie, tfidf_recs)
    semantic_score = genre_consistency(movie, semantic_recs)

    comparison_results.append({
        "Movie": movie,
        "TF-IDF Genre Consistency": tfidf_score,
        "Semantic Genre Consistency": semantic_score
    })

comparison_df = pd.DataFrame(comparison_results)

comparison_df
```

```python
import numpy as np
import pandas as pd

def get_hybrid_recommendations(title, n=10, alpha=0.5):
    matches = df[df["title"].str.lower() == title.lower()]

    if matches.empty:
        print(f"Movie '{title}' not found.")
        return pd.DataFrame()

    idx = matches.index[0]

    # Get TF-IDF similarity scores
    tfidf_scores = cosine_sim[idx].copy()

    # Get semantic similarity scores
    semantic_scores = semantic_similarity[idx].copy()

    # Normalize both scores to 0-1
    tfidf_min = tfidf_scores.min()
    tfidf_max = tfidf_scores.max()

    semantic_min = semantic_scores.min()
    semantic_max = semantic_scores.max()

    tfidf_norm = (
        (tfidf_scores - tfidf_min) /
        (tfidf_max - tfidf_min)
    )

    semantic_norm = (
        (semantic_scores - semantic_min) /
        (semantic_max - semantic_min)
    )

    # Combine the two models
    hybrid_scores = (
        alpha * tfidf_norm +
        (1 - alpha) * semantic_norm
    )

    # Create index-score pairs
    similarity_scores = list(
        enumerate(hybrid_scores)
    )

    # Sort by hybrid score
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Remove the movie itself
    similarity_scores = similarity_scores[1:n+1]

    movie_indices = [x[0] for x in similarity_scores]

    result = df.iloc[movie_indices][
        ["id", "title", "vote_average", "popularity"]
    ].copy()

    result["hybrid_score"] = [
        x[1] for x in similarity_scores
    ]

    return result.reset_index(drop=True)
```

```python
def precision_at_k(recommended_titles, relevant_titles, k=10):

    recommended = recommended_titles[:k]

    relevant = set(
        title.lower()
        for title in relevant_titles
    )

    hits = sum(
        title.lower() in relevant
        for title in recommended
    )

    return hits / k
```

```python
def recall_at_k(recommended_titles, relevant_titles, k=10):

    recommended = recommended_titles[:k]

    relevant = set(
        title.lower()
        for title in relevant_titles
    )

    hits = sum(
        title.lower() in relevant
        for title in recommended
    )

    if len(relevant) == 0:
        return 0

    return hits / len(relevant)
```

```python
import numpy as np

def ndcg_at_k(recommended_titles, relevant_titles, k=10):

    recommended = recommended_titles[:k]

    relevant = set(
        title.lower()
        for title in relevant_titles
    )

    relevance_scores = [
        1 if title.lower() in relevant else 0
        for title in recommended
    ]

    dcg = 0

    for i, score in enumerate(relevance_scores):
        dcg += score / np.log2(i + 2)

    ideal_scores = sorted(
        relevance_scores,
        reverse=True
    )

    idcg = 0

    for i, score in enumerate(ideal_scores):
        idcg += score / np.log2(i + 2)

    if idcg == 0:
        return 0

    return dcg / idcg
```

```python
evaluation_results = []

models = {
    "Semantic": 0.0,
    "Hybrid": 0.5,
    "TF-IDF": 1.0
}

for model_name, alpha in models.items():

    precision_scores = []
    recall_scores = []
    ndcg_scores = []

    for movie, relevant_movies in relevance_benchmark.items():

        recommendations = get_hybrid_recommendations(
            movie,
            n=10,
            alpha=alpha
        )

        recommended_titles = recommendations["title"].tolist()

        precision_scores.append(
            precision_at_k(
                recommended_titles,
                relevant_movies,
                k=10
            )
        )

        recall_scores.append(
            recall_at_k(
                recommended_titles,
                relevant_movies,
                k=10
            )
        )

        ndcg_scores.append(
            ndcg_at_k(
                recommended_titles,
                relevant_movies,
                k=10
            )
        )

    evaluation_results.append({
        "model": model_name,
        "precision@10": np.mean(precision_scores),
        "recall@10": np.mean(recall_scores),
        "ndcg@10": np.mean(ndcg_scores)
    })

ranking_evaluation = pd.DataFrame(evaluation_results)

ranking_evaluation
```
